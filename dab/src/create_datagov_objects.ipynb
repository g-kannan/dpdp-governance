```sql
create catalog if not exists datagov
```

```sql
CREATE OR REPLACE TABLE 
-- IF NOT EXISTS 
datagov.default.column_scan_metadata (
  table_catalog STRING,
  table_schema STRING,
  table_name STRING,
  column_name STRING,
  column_hash STRING,
  data_type STRING,
  classification_desc STRING,
  classification_code INT,
  is_identity_spine BOOLEAN,
  detection_method STRING,
  confidence_score DECIMAL(2,2),
  validation_status STRING,
  scanned_ts TIMESTAMP,
  tag_status STRING,
  tagged_ts TIMESTAMP,
  created_ts TIMESTAMP,
  updated_ts TIMESTAMP
  )
```

```sql
CREATE TABLE IF NOT EXISTS datagov.default.tag_errors (
    error_timestamp TIMESTAMP,
    failed_command STRING,
    error_message STRING
  )
```
